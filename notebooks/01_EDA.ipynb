# 4. Literature Review

## 4.1 Loan Eligibility Prediction

Public loan prediction datasets commonly include demographic information, applicant income, co-applicant income, requested loan amount, repayment term, credit history, and property area. These features provide a basis for studying applicant characteristics and practicing classification techniques.

The Kaggle Loan Eligibility Prediction dataset documentation describes loan approval as a classification task in which the target represents approved or rejected applications.

Reference: [Kaggle — Loan Eligibility Prediction](https://www.kaggle.com/competitions/loan-eligibility-prediction/data)

## 4.2 Importance of Exploratory Data Analysis

Exploratory Data Analysis helps analysts understand the distribution, variability, and relationships within a dataset before building a predictive model. Histograms, boxplots, scatterplots, countplots, and correlation heatmaps help reveal patterns, potential outliers, and data-quality problems.

The Pandas documentation provides guidance on data inspection, descriptive statistics, categorical data, and missing-value handling.

Reference: [Pandas User Guide](https://pandas.pydata.org/docs/user_guide/index.html)

## 4.3 Data Preprocessing and Machine Learning Pipelines

Machine learning datasets frequently contain missing values and categorical features that require preprocessing. Numerical and categorical columns may require different transformations.

Scikit-learn provides tools such as `SimpleImputer`, `StandardScaler`, `OneHotEncoder`, `ColumnTransformer`, and `Pipeline` to organize these steps. Fitting preprocessing transformations only on training data helps prevent data leakage.

Reference: [Scikit-learn — Getting Started](https://scikit-learn.org/stable/getting_started.html)

## 4.4 Classification Algorithms and Evaluation

Binary classification algorithms learn to assign observations to one of two target classes. Logistic Regression provides a baseline model, while Decision Trees, Random Forests, SVM, KNN, and Gradient Boosting offer alternative approaches to learning classification boundaries.

Model evaluation should not rely exclusively on accuracy. Precision, recall, F1-score, and confusion matrices provide additional information about different types of classification errors.

Reference: [Scikit-learn — Classification and Model Evaluation](https://scikit-learn.org/stable/modules/model_evaluation.html)

## 4.5 Research Gap and Project Motivation

The key practical challenge is not merely selecting a machine learning algorithm. It is understanding the dataset, identifying data-quality issues, choosing appropriate preprocessing methods, and evaluating results responsibly.

This project provides hands-on experience with these steps. However, because its target variable is synthetic, it cannot establish which applicant characteristics genuinely influence loan approvals or determine the real-world performance of a loan eligibility system.

# 5. What I Have Learned from the Literature Review

The literature review helped me understand the following concepts:

1. **EDA should precede modeling:** Data quality and distributions should be investigated before training algorithms.
2. **Missing values require deliberate treatment:** The appropriate strategy depends on the feature and the meaning of missingness.
3. **Outliers need investigation:** An unusual observation may be a valid record rather than an error.
4. **Different variables require different visualizations:** Histograms and boxplots are useful for numerical distributions, while countplots are useful for categorical frequencies.
5. **Relationships do not establish causation:** Correlation and grouped comparisons reveal associations, not proof that one variable causes another.
6. **Categorical features require suitable encoding:** Algorithms generally need categorical inputs represented numerically.
7. **Data leakage must be avoided:** Preprocessing should be fitted using training data rather than the full dataset before evaluation.
8. **Multiple evaluation metrics are important:** Accuracy alone may hide poor performance on one of the classes.
9. **Model comparison requires a consistent evaluation setup:** Algorithms should use the same train-test split and evaluation metrics.
10. **Target quality determines what a model can learn:** A model trained on synthetic labels learns patterns created by the synthetic-label generation process, not necessarily genuine real-world relationships.

# Univariate Analysis

## Numerical columns

```python
numerical_columns = [
    "ApplicantIncome",
    "CoapplicantIncome",
    "LoanAmount",
    "Loan_Amount_Term",
]

for column in numerical_columns:
    plt.figure(figsize=(7, 4))

    sns.histplot(data=df, x=column, kde=True, bins=30)

    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")

    plt.tight_layout()
    plt.show()
```

### Insights 

#### Applicant income distribution:
Most of the applicants have incomes closer to the lower end of the distribution, while a small number of applicants have very high incomes.  So the applicant income distribution appears to be right-skewed. 

#### Coappliant Income distribution:
Coapplicat inome varies considerably across applications, and zero income observations are an important part of the dataset. Mean coapplicant income is approx. 1570, while the median is 1025. The difference suggests that higher income observations are pulling the mean upward.

#### Loan Amount Distribution:
Loan amounts vary across applicants, with some relatively large requets potentially influencing overall distributions. The median loan amount is 125, while the mean approx. 136. The mean being higher than median suggests a right skewed distribution. 

#### Loan Amount Term Distribution:
The median loan term is 360 months, equivalent to 30 years. The central value of repayment period is relatively long. The dataset also contains shorter and longer loan terms, with observed values ranging from 6 to 480 months. so this tells us repayment periods are not identical for all applicants.

```python
for column in numerical_columns:
    plt.figure(figsize=(7, 3))

    sns.boxplot(data=df, x=column)

    plt.title(f"Boxplot of {column}")
    plt.xlabel(column)

    plt.tight_layout()
    plt.show()
```

### Insights

#### Boxplot :
The boxplots help identify observations that lie unusually far from the central range of the numerical variables. Applicant income and coapplicant income contain particularly large maximum values compared with their medians.

## Categorical columns

```python
categorical_columns = [
    "Gender",
    "Married",
    "Dependents",
    "Education",
    "Self_Employed",
    "Credit_History",
    "Property_Area",
    "Loan_Status",
]

for column in categorical_columns:
    plt.figure(figsize=(7, 4))

    sns.countplot(data=df, x=column)

    plt.title(f"Countplot of {column}")
    plt.xlabel(column)
    plt.ylabel("Number of Applicants")

    plt.tight_layout()
    plt.show()
```

### Insights

#### Gender Distribution:
The dataset contains 286 male applicants and 70 female applicants. There are also 11 missing gender values. Male applicants therefore make up the larger share of the recorded gender categories. 

#### Marital Status Distribution:
The dataset contains 233 applicants recorded as married and 134 recorded as unmarried. Married applicants represent the larger group. But it doesn't establish that martial status affects loan approval.

#### Dependents Distribution:
The largest category is applicants with zero dependents, containing 200 records. The remaining applicants are distributed across the categories of one dependent, two dependents, and three or more dependents. The 3+ category is grouped rather than representing a exact count.

#### Education Distribution:
The dataset contains 283 graduates and 84 non-graduates. Graduates make up the larger proportion of applicants. But it doesn't indicate any relationship between education and loan approval.

#### Self Employment Distribution:
The dataset contains 307 applicants recorded as not self employed and 37 recorded as self-employed . So most of the applicants are not self employed.

#### Credit History Distribution:
Among the recorded credit history values, 279 applicants have a value of 1 and 59 have a value of 0. The value  1  is more common than value 0 . Credit history value 1 means applicant's credit history is good and value 0 means applicant's credit history is poor and doesn't meet criteria.

#### Property Area Distribution:
The dataset contains 140 urban, 116 semiurban and 111 rural applicants. Urban applicants form largest category, but the difference between the three groups are relatively modest.

#### Loan Status Distribution:
The dataset contains 283 records labelled Y and 84 records labelled N. Approx. 77.1% of the records belong to class Y, while 22.9 belong to class N. And These lables were synthetically generated rather than obtained from actual bank decisions.

## Loan Status by category

```python
status_percentage = pd.crosstab(
    df["Education"],
    df["Loan_Status"],
    normalize="index"
) * 100

print(status_percentage.round(2))

status_percentage.plot(
    kind="bar",
    stacked=True,
    figsize=(7, 4)
)

plt.title("Loan Status Percentage by Education")
plt.xlabel("Education")
plt.ylabel("Percentage")
plt.legend(title="Loan Status")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

#### Insights:
The normalized stacked bar chart compares the percentage of Y and N labels within each education category rather than comparing raw counts alone. The Y-label proportion is approximately 77.4% among graduates and 76.2% among non-graduates. These proportions are close, suggesting that the generated target distribution is similar across the two education groups.
